**ML**

```python
x=data.select_dtypes(include="number").drop(columns=["price_usd"])
y=data["price_usd"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_squared_error
```

```python
model=LinearRegression()
model.fit(x_train_scaled,y_train)
```

```python
y_pred=model.predict(x_test_scaled)
```

```python
print(r2_score(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```

```python
plt.figure(figsize=(10,8))
plt.scatter(y_test,y_pred,color="purple")
plt.xlabel("Actural")
plt.ylabel("Predicted")
plt.title("Actual Vs Predicted")
plt.show()
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
gbr=GradientBoostingRegressor()
gbr.fit(x_train,y_train)
```

```python
y_pre=gbr.predict(x_test)
```

```python
print("R2:", r2_score(y_test, y_pre))
print("MAE:", mean_absolute_error(y_test, y_pre))
print("MSE:", mean_squared_error(y_test, y_pre))
```

```python
plt.figure(figsize=(10,8))
plt.scatter(y_test,y_pre,color="green")
plt.xlabel("Actural")
plt.ylabel("Predicted")
plt.title("Actual Vs Predicted")
plt.show()
```

```python
models=["LinearRegression","Gradient Boosting"]
scores=[0.75,0.82]

plt.figure(figsize=(8,6))
plt.bar(models,scores,color=["green","yellow"])
plt.ylabel("R2 Scores")
plt.title("Model Comparison",fontsize=15,color="red")

for i, v in enumerate(scores):
  plt.text(i,v,str(round(v,2)),ha="center",fontsize=14)

plt.show()
```
